```python
sales_data = [
    {"rep": "Alice", "region": "North", "month": "Jan", "target": 50000, "achieved": 54000},
    {"rep": "Bob", "region": "South", "month": "Jan", "target": 45000, "achieved": 41000},
    {"rep": "Carol", "region": "East", "month": "Jan", "target": 60000, "achieved": 62000},
    {"rep": "David", "region": "North", "month": "Feb", "target": 50000, "achieved": 48000},
    {"rep": "Alice", "region": "North", "month": "Feb", "target": 55000, "achieved": 57000},
    {"rep": "Bob", "region": "South", "month": "Feb", "target": 45000, "achieved": 50000},
    {"rep": "Carol", "region": "East", "month": "Feb", "target": 60000, "achieved": 58000},
    {"rep": "David", "region": "North", "month": "Jan", "target": 40000, "achieved": 39000},
]
total_achieved = 0
met_target = []
missed_target = []
rep_totals = {}
north_total =0

for record in sales_data:
    total_achieved += record["achieved"]

    if record["achieved"] >= record["target"]:
        met_target.append(record["rep"])
    else:
        missed_target.append(record["rep"])

    rep = record["rep"]
    rep_totals[rep] = rep_totals.get(rep, 0) + record["achieved"]

    if record["region"] == "North":
        north_total += record["achieved"]

best_rep = max(rep_totals, key=rep_totals.get)

print("Total Achieved:", total_achieved)
print("Met Target:", met_target)
print("Missed Target:", missed_target)
print("Best Performing Rep:", best_rep)
print("North Region Total Achieved:", north_total)
```
